```python
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
email_data = {
    "message": [
        "Congratulations! You won a free vacation. Click here to claim your prize.",
        "Please review the notes from today's department meeting.",
        "Earn money quickly by working from home. Sign up today.",
        "Your assignment is due Sunday at 11:59 p.m.",
        "You have been selected to receive a free gift card.",
        "Can we schedule a meeting for tomorrow afternoon?",
        "Urgent! Confirm your bank information to avoid account closure.",
        "Thank you for submitting your project.",
        "Limited-time offer! Buy now and receive 50 percent off.",
        "Here is the presentation we discussed during class.",
        "Claim your cash reward before the offer expires.",
        "The course materials have been posted in Blackboard.",
        "You are a winner! Respond now to collect your prize.",
        "Please send me your availability for next week's meeting.",
        "Get approved for a loan today with no credit check.",
        "Your order has shipped and will arrive on Friday.",
        "Click this link to receive a free smartphone.",
        "The updated report is attached for your review.",
        "Act now to receive an exclusive financial reward.",
        "Please remember to complete the course survey."
    ],
    "label": [
        "spam",
        "not spam",
        "spam",
        "not spam",
        "spam",
        "not spam",
        "spam",
        "not spam",
        "spam",
        "not spam",
        "spam",
        "not spam",
        "spam",
        "not spam",
        "spam",
        "not spam",
        "spam",
        "not spam",
        "spam",
        "not spam"
    ]
}

df = pd.DataFrame(email_data)

df
```

```python
df["label"].value_counts()
```

```python
X = df["message"]
y = df["label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Training examples:", len(X_train))
print("Testing examples:", len(X_test))
```

```python
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english"
)

X_train_vectors = vectorizer.fit_transform(X_train)
X_test_vectors = vectorizer.transform(X_test)
```

```python
model = MultinomialNB()

model.fit(X_train_vectors, y_train)
```

```python
predictions = model.predict(X_test_vectors)

results = pd.DataFrame({
    "Email Message": X_test,
    "Actual Label": y_test,
    "Predicted Label": predictions
})

results
```

```python
accuracy = accuracy_score(y_test, predictions)

print("Model accuracy:", accuracy)
print()
print(classification_report(y_test, predictions, zero_division=0))
```

```python
new_emails = [
    "Congratulations! Click here to collect your free cash prize.",
    "Please review the attached agenda before tomorrow's meeting.",
    "You have been selected for a special offer. Respond immediately.",
    "Your instructor has posted feedback on your assignment."
]

new_email_vectors = vectorizer.transform(new_emails)
new_predictions = model.predict(new_email_vectors)

for email, prediction in zip(new_emails, new_predictions):
    print("Email:", email)
    print("Prediction:", prediction)
    print()
```
